```python
import yfinance as yf

# 투자금
investment = 10_000_000  # 1,000만원

# 엔비디아 2020년 주가 데이터
nvda = yf.download(
    "NVDA",
    start="2020-01-01",
    end="2026-09-21",
    auto_adjust=True
)

# 2020년 첫 거래일 / 마지막 거래일 종가
start_price = float(nvda["Close"].iloc[0])
end_price = float(nvda["Close"].iloc[-1])

# 주가 상승률
return_rate = (end_price - start_price) / start_price

# 최종 금액
final_money = investment * (1 + return_rate)

# 수익금
profit = final_money - investment

print(f"2020년 시작 주가: ${start_price:.2f}")
print(f"2020년 마지막 주가: ${end_price:.2f}")
print(f"수익률: {return_rate * 100:.2f}%")
print(f"투자금: {investment:,.0f}원")
print(f"최종 금액: {final_money:,.0f}원")
print(f"수익금: {profit:,.0f}원")
```
